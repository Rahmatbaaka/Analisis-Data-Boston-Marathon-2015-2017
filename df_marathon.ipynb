## Import semua lirary/packages yang di gunakan

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Data Wrangling

### Gathering Data

Menyiapkan Dataset

```python
df_marathon2015 = pd.read_csv("https://raw.githubusercontent.com/Rahmatbaaka/Analisis-Data-Boston-Marathon-2015-2017/main/data/marathon_results_2015.csv")
df_marathon2015.head()
```

```python
df_marathon2016 = pd.read_csv("https://raw.githubusercontent.com/Rahmatbaaka/Analisis-Data-Boston-Marathon-2015-2017/main/data/marathon_results_2016.csv")
df_marathon2016.head()
```

```python
df_marathon2017 = pd.read_csv("https://raw.githubusercontent.com/Rahmatbaaka/Analisis-Data-Boston-Marathon-2015-2017/main/data/marathon_results_2017.csv")
df_marathon2017.head()
```

karena kolom dan isi darii ketiga data sama, maka akan kita gabung menjadi 1 data yang akan memudahkan analisis kita. sebelumnya akan kita tambahkan kolom year untuk membedakan tahun nya

```python
df_marathon2015["year"] = 2015
df_marathon2016["year"] = 2016
df_marathon2017["year"] = 2017
```

```python
df_marathon = pd.concat([df_marathon2015, df_marathon2016, df_marathon2017])
df_marathon
```

### Cleaning Data

Menghapus kolom yang tidak diperlukan

```python
df_marathon = df_marathon.drop(["Unnamed: 0",	"Bib", "Name", "City", "State", "Citizen", "5K", "10K", "15K", "20K", "Half", "25K", "30K", "35K", "40K", "Proj Time", "Unnamed: 9", "Unnamed: 8", "Division", "Gender"], axis=1)
```

```python
df_marathon
```

mengganti nama kolom M/F dengan Gender

```python
df_marathon.rename(columns={"M/F": "Gender"}, inplace=True)
```

```python
df_marathon.info()
```

Mengganti type data beberapa kolom untuk memudahkan analisis

```python
df_marathon["Official Time"] = pd.to_timedelta(df_marathon["Official Time"])

df_marathon["Pace"] = df_marathon["Pace"].str.split(":").apply(lambda x: int(x[0]) + int(x[1])/60)

df_marathon["Gender"] = df_marathon["Gender"].astype("category")
```

menambah kolom Time Minutes dimana merupakan data dari Official Time yang di jadikan float berupa menit, agar mudah dalam analisis data

```python
df_marathon["Time Minutes"] = df_marathon["Official Time"].dt.total_seconds() / 60
```

```python
df_marathon.info()
```

## Exploratory Data Analysis (EDA)

analisis dasar

```python
df_marathon[["Age","Pace","Time Minutes"]].describe()
```

Rata-rata per tahun

```python
df_marathon.groupby("year")[["Age","Pace","Time Minutes"]].mean()
```

distribusi umur

```python
plt.figure(figsize=(8,5))
sns.histplot(df_marathon["Age"], bins=30)
plt.title("Distribusi Usia Pelari")
plt.show()
```

distribusi pace

```python
plt.figure(figsize=(8,5))
sns.histplot(df_marathon["Pace"], bins=30)
plt.title("Distribusi Pace")
plt.show()
```

Distribsi waktu

```python
plt.figure(figsize=(8,5))
sns.histplot(df_marathon["Time Minutes"], bins=30)
plt.title("Distribusi Waktu Finish")
plt.show()
```

2. Analisis Hubungan Variabel

umur vs pace

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="Age", y="Pace", data=df_marathon, alpha=0.4)
plt.title("Hubungan Usia vs Pace")
plt.show()
```

Gender vs Performa

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="Gender", y="Time Minutes", data=df_marathon)
plt.title("Performa Gender vs Waktu Finish")
plt.show()
```

top 10 negara pelari terbanyak

```python
top_country = df_marathon["Country"].value_counts().head(10)

plt.figure(figsize=(10,6))
top_country.plot(kind="bar")
plt.title("Top 10 Negara Peserta Marathon")
plt.show()
```

top 10 negara tercepat

```python
fastest_country = df_marathon.groupby("Country")["Time Minutes"].mean().nsmallest(10)

plt.figure(figsize=(10,6))
fastest_country.plot(kind="bar")
plt.title("Negara dengan Rata-rata Finish Tercepat")
plt.show()
```

3. Deteksi Pola dan Trend

Tren waktu tiap tahun

```python
trend_time = df_marathon.groupby("year")["Time Minutes"].mean()

plt.figure(figsize=(8,5))
trend_time.plot(marker="o")
plt.title("Tren Waktu Finish per Tahun")
plt.ylabel("Menit")
plt.show()
```

Tren Pace tiap tahun

```python
trend_pace = df_marathon.groupby("year")["Pace"].mean()

plt.figure(figsize=(8,5))
trend_pace.plot(marker="o")
plt.title("Tren Pace per Tahun")
plt.show()
```

Perubahan performa berdasarkan umur

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="Age" , y="Time Minutes", data=df_marathon)
plt.xticks(rotation=90)
plt.title("Performa Berdasarkan Usia")
plt.show()
```

```python
min_age = df_marathon['Age'].min()
max_age = df_marathon['Age'].max()

bins = range(int(min_age), int(max_age) + 5, 5)

labels = [f"{i}-{i+4}" for i in bins[:-1]]

df_marathon['Age_Group'] = pd.cut(df_marathon['Age'], bins=bins, labels=labels)

df_marathon = df_marathon.reset_index(drop=True)

plt.figure(figsize=(14,6))

sns.boxplot(
    x="Age_Group",
    y="Time Minutes",
    data=df_marathon,
    showfliers=False  # supaya tidak terlalu ramai
)

plt.title("Performa Marathon Berdasarkan Kelompok Usia (Interval 5 Tahun)", fontsize=14)
plt.xlabel("Kelompok Usia")
plt.ylabel("Waktu Finish (Menit)")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

Sebaran pelari tiap tahun

```python
plt.figure(figsize=(8,5))
df_marathon["year"].value_counts().sort_index().plot(kind="bar")
plt.xticks(rotation=0)
plt.title("Jumlah Peserta per Tahun")
plt.show()
```
